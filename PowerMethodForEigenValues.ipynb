```python
def ludecompose(a):
    a=a.astype(float)
    n= len(a)
    L= np. zeros((n,n))
    #t.tic()
    for k in range(n-1):
        p=k 
        big=abs(a[k,k])
        for i in range(k+1,n):
            if (a[i,k]>big):
                big= abs(a[i,k])
                p=i
            if (p != k):
                a[[k,p]] = a[[p,k]]
            factor= a[i,k]/a[k,k]
            L[i,k]= factor 
            for j in range (k,n):
                a[i,j] -= factor * a[k,j]

    for i in range (n):
        L[i,i] = 1
    #t.toc("Time spent in elimination is")
    u=a 
    #print(L@u)
    return L,u
```

```python
def forwardsub(L, b):
    n = len(b)
    y = np.zeros(n)
    #t.tic()
    for i in range(n):                   
        tot = b[i]
        for j in range(i):      
            tot = tot - L[i,j] * y[j]
        y[i] = tot / L[i,i]
    #t.toc("Forward Sub time" )
    return y
```

```python
def backsub(u,y):
    n = len(y)
    x = np.zeros(n)
    #t.tic()
    for i in range(n-1, -1, -1):
        tot = y[i]
        for j in range(i+1, n):
            tot = tot - u[i,j] * x[j]
        x[i] = tot / u[i,i]
    #t.toc()
    return x
```
