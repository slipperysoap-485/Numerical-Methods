```python
def backsub(a,b):
    n= len(b)
    x= np.zeros(n)
    for i in range (n-1,-1,-1):
        tot=b[i]
        for j in range (i+1,n):
            tot= tot - a[i,j] * x[j] 
        x[i] =  tot / a[i,i]
    return x
    
```
